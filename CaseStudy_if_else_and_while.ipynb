This app will accept student names and GPAs and test if the student qualifies for either the Dean's List or the Honor Roll. 

```python
#get last name as input and test whether it's 'ZZZ'
last_name = input("Enter your last name: ")
first_name = input("Enter your first name: ")

if last_name == "zzz":
   print("EOF")
else:
   gpa = input("Enter your G.P.A.: ")
   print(f"Hello,{first_name} {last_name}! Your G.P.A. is {gpa}")
```
